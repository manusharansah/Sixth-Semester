## Q1. Program to Implement GCD using Euclid's Algorithm

### Related Theory
The **Greatest Common Divisor (GCD)** of two integers is the largest positive integer that divides both numbers without leaving a remainder. **Euclid's Algorithm** is an efficient iterative method based on the property:

$$\gcd(a, b) = \gcd(b, a \bmod b), \quad \text{with } \gcd(a, 0) = a$$

The algorithm repeatedly replaces the larger number by the remainder of dividing the larger by the smaller, until the remainder becomes zero. The last non-zero remainder is the GCD.

### Related Diagram (Flow)
```
        Start
          |
       Read a, b
          |
   +------------------+
   | while b != 0:    | <---+
   |   r = a % b      |     |
   |   a = b          |     |
   |   b = r          | ----+
   +------------------+
          |
   Output a  (GCD)
          |
         End
```


````python
def gcd_euclid(a, b):
    """Return GCD of a and b using Euclid's Iterative Algorithm."""
    a, b = abs(a), abs(b)
    while b != 0:
        a, b = b, a % b
    return a


# Driver / Test
if __name__ == "__main__":
    num1, num2 = 252, 105
    print(f"GCD of {num1} and {num2} is: {gcd_euclid(num1, num2)}")

    num1, num2 = 1071, 462
    print(f"GCD of {num1} and {num2} is: {gcd_euclid(num1, num2)}")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(\log(\min(a, b)))$ — Euclid's algorithm reduces the problem size very quickly (related to the Fibonacci ratio), giving logarithmic performance instead of the linear performance of naive subtraction-based GCD.
- **Space Complexity:** $O(1)$ — only a constant number of variables are used (iterative, not recursive).
- **Remark:** This is one of the oldest known algorithms (circa 300 BC) and is optimal for GCD computation of two integers.

## Q2. Program to Implement Fibonacci Series

### Related Theory
The **Fibonacci sequence** is a series of numbers where each number is the sum of the two preceding numbers, starting from 0 and 1:

$$F_0 = 0,\quad F_1 = 1,\quad F_n = F_{n-1} + F_{n-2} \; (n \ge 2)$$

It can be generated **iteratively** using a loop, which avoids the exponential time cost of the naive recursive approach.

### Related Diagram (Flow)
```
        Start
          |
   Read n (terms)
          |
   a = 0, b = 1
          |
   +---------------------+
   | for i in range(n):  | <---+
   |   print(a)          |     |
   |   a, b = b, a+b     | ----+
   +---------------------+
          |
         End
```


````python
def fibonacci_iterative(n):
    """Return a list containing the first n terms of the Fibonacci series."""
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series


# Driver / Test
if __name__ == "__main__":
    n_terms = 12
    result = fibonacci_iterative(n_terms)
    print(f"First {n_terms} terms of Fibonacci series:")
    print(result)
````

### Analysis of the Algorithm
- **Time Complexity:** $O(n)$ — a single loop runs `n` times, each iteration doing constant work.
- **Space Complexity:** $O(n)$ for storing the series (or $O(1)$ if only the n-th term is required, without storing the whole list).
- **Remark:** The iterative approach is far more efficient than the naive recursive definition, which has exponential time complexity $O(2^n)$ due to repeated recomputation of the same sub-problems.

## Q3. Program to Implement Sequential (Linear) Search

### Related Theory
**Sequential Search** (or Linear Search) is the simplest searching technique. It examines each element of the list one by one, starting from the first element, and compares it with the target (key) value until a match is found or the list is exhausted. It does not require the list to be sorted.

### Related Diagram (Flow)
```
        Start
          |
   Read list, key
          |
   i = 0
          |
   +---------------------------+
   | while i < len(list):      | <---+
   |   if list[i] == key:      |     |
   |      return i             |     |
   |   i = i + 1               | ----+
   +---------------------------+
          |
   return -1 (not found)
          |
        End
```


````python
def sequential_search(arr, key):
    """Return the index of key in arr using sequential search, or -1 if not found."""
    for index in range(len(arr)):
        if arr[index] == key:
            return index
    return -1


# Driver / Test
if __name__ == "__main__":
    data = [45, 12, 78, 34, 90, 23, 56]
    target = 34
    pos = sequential_search(data, target)
    if pos != -1:
        print(f"Element {target} found at index {pos}")
    else:
        print(f"Element {target} not found in the list")

    target2 = 100
    pos2 = sequential_search(data, target2)
    print(f"Search result for {target2}: {'Found at index ' + str(pos2) if pos2 != -1 else 'Not Found'}")
````

### Analysis of the Algorithm
- **Time Complexity:**
  - Best Case: $O(1)$ — element found at the first position.
  - Worst / Average Case: $O(n)$ — element found at the last position or not present at all.
- **Space Complexity:** $O(1)$ — no extra space required.
- **Remark:** Sequential search works on both sorted and unsorted data but is inefficient for large datasets compared to Binary Search (which requires sorted data).

## Q4. Program to Implement Bubble Sort

### Related Theory
**Bubble Sort** is a simple comparison-based sorting algorithm. It repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. This process is repeated until no swaps are needed, meaning the list is sorted. Larger elements "bubble" toward the end of the list with each pass.

### Related Diagram (Flow)
```
        Start
          |
   Read list of n elements
          |
   +---------------------------------+
   | for i in range(n-1):            | <---+
   |   for j in range(0, n-i-1):     |     |
   |     if arr[j] > arr[j+1]:       |     |
   |        swap(arr[j], arr[j+1])   |     |
   |                                 | ----+
   +---------------------------------+
          |
   Output sorted list
          |
        End
```


````python
def bubble_sort(arr):
    """Sort a list in ascending order using Bubble Sort (returns a new sorted list)."""
    a = arr.copy()
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        if not swapped:      # optimization: stop early if already sorted
            break
    return a


# Driver / Test
if __name__ == "__main__":
    numbers = [64, 34, 25, 12, 22, 11, 90]
    print("Original list:", numbers)
    sorted_numbers = bubble_sort(numbers)
    print("Sorted list  :", sorted_numbers)
````

### Analysis of the Algorithm
- **Time Complexity:**
  - Worst / Average Case: $O(n^2)$ — nested loops compare every pair of adjacent elements.
  - Best Case: $O(n)$ — when the list is already sorted (with the "swapped" early-exit optimization).
- **Space Complexity:** $O(1)$ — sorting is done in-place (auxiliary array used here only to preserve the original input).
- **Remark:** Bubble Sort is stable and simple to implement, but inefficient for large datasets compared to $O(n\log n)$ algorithms like Merge Sort or Quick Sort.

## Q5. Program to Implement Insertion Sort

### Related Theory
**Insertion Sort** builds the sorted list one element at a time. It takes each element from the unsorted part and inserts it into its correct position within the already-sorted part of the list, shifting larger elements to the right as necessary. It resembles the way people sort playing cards in their hands.

### Related Diagram (Flow)
```
        Start
          |
   Read list of n elements
          |
   +--------------------------------------+
   | for i in range(1, n):                | <---+
   |   key = arr[i]                       |     |
   |   j = i - 1                          |     |
   |   while j >= 0 and arr[j] > key:     |     |
   |       arr[j+1] = arr[j]; j = j - 1   |     |
   |   arr[j+1] = key                     | ----+
   +--------------------------------------+
          |
   Output sorted list
          |
        End
```


````python
def insertion_sort(arr):
    """Sort a list in ascending order using Insertion Sort (returns a new sorted list)."""
    a = arr.copy()
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a


# Driver / Test
if __name__ == "__main__":
    numbers = [12, 11, 13, 5, 6]
    print("Original list:", numbers)
    sorted_numbers = insertion_sort(numbers)
    print("Sorted list  :", sorted_numbers)
````

### Analysis of the Algorithm
- **Time Complexity:**
  - Worst / Average Case: $O(n^2)$ — each new element may need to be compared/shifted against all previously sorted elements.
  - Best Case: $O(n)$ — when the list is already sorted, only one comparison per element is needed.
- **Space Complexity:** $O(1)$ — sorting is done in-place.
- **Remark:** Insertion Sort is efficient for small or nearly-sorted datasets and is used as part of hybrid algorithms like Timsort.

## Q6. Program to Implement Selection Sort

### Related Theory
**Selection Sort** divides the list into a sorted and an unsorted part. In every pass, it finds the **minimum element** from the unsorted part and swaps it with the first element of the unsorted part, thereby growing the sorted part by one element each time.

### Related Diagram (Flow)
```
        Start
          |
   Read list of n elements
          |
   +----------------------------------------+
   | for i in range(n-1):                   | <---+
   |   min_idx = i                          |     |
   |   for j in range(i+1, n):              |     |
   |       if arr[j] < arr[min_idx]:        |     |
   |            min_idx = j                 |     |
   |   swap(arr[i], arr[min_idx])           | ----+
   +----------------------------------------+
          |
   Output sorted list
          |
        End
```


````python
def selection_sort(arr):
    """Sort a list in ascending order using Selection Sort (returns a new sorted list)."""
    a = arr.copy()
    n = len(a)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]
    return a


# Driver / Test
if __name__ == "__main__":
    numbers = [64, 25, 12, 22, 11]
    print("Original list:", numbers)
    sorted_numbers = selection_sort(numbers)
    print("Sorted list  :", sorted_numbers)
````

### Analysis of the Algorithm
- **Time Complexity:** $O(n^2)$ in the best, average, and worst cases — the algorithm always scans the remaining unsorted portion to find the minimum, regardless of the initial order.
- **Space Complexity:** $O(1)$ — sorting is done in-place.
- **Remark:** Selection Sort performs the minimum possible number of swaps ($n-1$), which makes it useful when write/swap operations are costly, even though comparisons remain $O(n^2)$.

## Q7. Program to Implement the Chinese Remainder Theorem (CRT)

### Related Theory
The **Chinese Remainder Theorem** states that if $n_1, n_2, \dots, n_k$ are **pairwise coprime** positive integers, then for any given remainders $a_1, a_2, \dots, a_k$, the system of simultaneous congruences

$$x \equiv a_1 \pmod{n_1}, \quad x \equiv a_2 \pmod{n_2}, \quad \dots, \quad x \equiv a_k \pmod{n_k}$$

has a **unique solution modulo** $N = n_1 \times n_2 \times \cdots \times n_k$.

The solution is computed as:
$$x = \left(\sum_{i=1}^{k} a_i \cdot M_i \cdot y_i \right) \bmod N$$
where $M_i = N / n_i$ and $y_i$ is the modular inverse of $M_i$ modulo $n_i$ (found using the Extended Euclidean Algorithm).

### Related Diagram (Flow)
```
        Start
          |
   Read remainders[], moduli[]
          |
   N = product(moduli)
          |
   +--------------------------------------------+
   | for each (a_i, n_i):                       | <---+
   |    M_i = N / n_i                           |     |
   |    y_i = modular_inverse(M_i, n_i)         |     |
   |    sum += a_i * M_i * y_i                  | ----+
   +--------------------------------------------+
          |
   x = sum mod N
          |
       Output x
          |
        End
```


````python
def extended_gcd(a, b):
    """Return (g, x, y) such that a*x + b*y = g = gcd(a, b)."""
    old_r, r = a, b
    old_s, s = 1, 0
    old_t, t = 0, 1
    while r != 0:
        q = old_r // r
        old_r, r = r, old_r - q * r
        old_s, s = s, old_s - q * s
        old_t, t = t, old_t - q * t
    return old_r, old_s, old_t   # g, x, y


def mod_inverse(a, m):
    g, x, _ = extended_gcd(a, m)
    if g != 1:
        raise ValueError(f"Modular inverse does not exist for {a} mod {m}")
    return x % m


def chinese_remainder_theorem(remainders, moduli):
    """Solve x = remainders[i] (mod moduli[i]) for all i, using CRT."""
    N = 1
    for n in moduli:
        N *= n

    x = 0
    for a_i, n_i in zip(remainders, moduli):
        M_i = N // n_i
        y_i = mod_inverse(M_i, n_i)
        x += a_i * M_i * y_i

    return x % N, N


# Driver / Test
if __name__ == "__main__":
    # Example: x = 2 (mod 3), x = 3 (mod 5), x = 2 (mod 7)
    remainders = [2, 3, 2]
    moduli = [3, 5, 7]

    solution, N = chinese_remainder_theorem(remainders, moduli)
    print(f"The system of congruences has solution x = {solution} (mod {N})")

    # Verification
    for a, n in zip(remainders, moduli):
        print(f"Check: {solution} mod {n} = {solution % n}  (expected {a})")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(k \log(\max(n_i)))$, where $k$ is the number of congruences — dominated by computing modular inverses via the Extended Euclidean Algorithm for each modulus.
- **Space Complexity:** $O(1)$ (excluding input storage).
- **Remark:** CRT requires the moduli to be **pairwise coprime**; it is widely used in cryptography (e.g., RSA decryption speed-up) and computer arithmetic for large-number computations.

## Q8. Program to Implement Fermat's (Little) Theorem

### Related Theory
**Fermat's Little Theorem** states that if $p$ is a prime number and $a$ is an integer not divisible by $p$, then:

$$a^{p-1} \equiv 1 \pmod p$$

This theorem forms the basis of the **Fermat Primality Test**: to check whether a number $p$ is (probably) prime, we pick a random base $a$ ($1 < a < p-1$) and check whether $a^{p-1} \bmod p = 1$. If this fails for some $a$, $p$ is definitely composite. If it holds for several random bases, $p$ is *probably* prime (a small chance of error exists for **Carmichael numbers**).

### Related Diagram (Flow)
```
        Start
          |
   Read p, number of trials k
          |
   +---------------------------------------------+
   | for i in range(k):                          | <---+
   |    pick random a in [2, p-2]                |     |
   |    if pow(a, p-1, p) != 1:                  |     |
   |         return "Composite"                  |     |
   |                                             | ----+
   +---------------------------------------------+
          |
   return "Probably Prime"
          |
        End
```


````python
import random

def fermat_primality_test(p, k=5):
    """Return True if p is 'probably prime' (Fermat test with k random bases)."""
    if p in (2, 3):
        return True
    if p < 2 or p % 2 == 0:
        return False

    for _ in range(k):
        a = random.randint(2, p - 2)
        if pow(a, p - 1, p) != 1:      # Fermat's Little Theorem check
            return False                 # Definitely composite
    return True                          # Probably prime


# Driver / Test
if __name__ == "__main__":
    test_numbers = [7, 15, 97, 561, 100, 104729]  # 561 is a Carmichael number
    for n in test_numbers:
        result = fermat_primality_test(n, k=10)
        print(f"{n:>7} -> {'Probably Prime' if result else 'Composite'}")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(k \log p)$ — each of the $k$ trials computes modular exponentiation using fast exponentiation, which takes $O(\log p)$ time.
- **Space Complexity:** $O(1)$.
- **Remark:** The Fermat test is a fast **probabilistic** primality test. Its main weakness is **Carmichael numbers** (e.g., 561), which satisfy $a^{p-1} \equiv 1 \pmod p$ for most bases $a$ despite being composite, causing false positives. This weakness motivates stronger tests such as **Miller–Rabin**.

## Q9. Program to Implement the Miller–Rabin Primality Test

### Related Theory
The **Miller–Rabin test** is a stronger probabilistic primality test that overcomes the Carmichael-number weakness of Fermat's test. Given an odd number $n$, we write:

$$n - 1 = 2^s \times d, \quad \text{where } d \text{ is odd}$$

For a randomly chosen base $a$, we compute $x = a^d \bmod n$. If $x = 1$ or $x = n-1$, $n$ passes this round. Otherwise, we repeatedly square $x$ up to $s-1$ times, checking whether it becomes $n-1$. If it never does, $n$ is **definitely composite** for that base. Repeating this test for several bases makes the probability of error extremely small.

### Related Diagram (Flow)
```
        Start
          |
   Read n, number of trials k
          |
   Write n-1 = 2^s * d  (d odd)
          |
   +---------------------------------------------------+
   | for i in range(k):                                | <---+
   |    pick random a in [2, n-2]                      |     |
   |    x = pow(a, d, n)                               |     |
   |    if x == 1 or x == n-1: continue (next trial)   |     |
   |    repeat s-1 times:                              |     |
   |        x = x*x mod n                              |     |
   |        if x == n-1: break (continue outer loop)   |     |
   |    else: return "Composite"                       | ----+
   +---------------------------------------------------+
          |
   return "Probably Prime"
          |
        End
```


````python
import random

def miller_rabin(n, k=5):
    """Return True if n is 'probably prime' using the Miller-Rabin test with k rounds."""
    if n in (2, 3):
        return True
    if n < 2 or n % 2 == 0:
        return False

    # Write n - 1 as 2^s * d  with d odd
    s, d = 0, n - 1
    while d % 2 == 0:
        d //= 2
        s += 1

    for _ in range(k):
        a = random.randint(2, n - 2)
        x = pow(a, d, n)
        if x == 1 or x == n - 1:
            continue                     # passes this round, try next base

        for _ in range(s - 1):
            x = pow(x, 2, n)
            if x == n - 1:
                break
        else:
            return False                  # definitely composite

    return True                            # probably prime


# Driver / Test
if __name__ == "__main__":
    test_numbers = [7, 15, 97, 561, 100, 104729, 1000000007]
    for n in test_numbers:
        result = miller_rabin(n, k=20)
        print(f"{n:>12} -> {'Probably Prime' if result else 'Composite'}")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(k \log^3 n)$ (or $O(k \log n)$ multiplications, each costing $O(\log^2 n)$ using fast modular exponentiation) — where $k$ is the number of rounds/bases tested.
- **Space Complexity:** $O(1)$.
- **Remark:** Miller–Rabin is far more reliable than Fermat's test — there are **no Carmichael-number-like exceptions** that fool it consistently, and the probability of a composite number being incorrectly reported as prime is at most $4^{-k}$. It is the primality test used in most practical cryptographic libraries (e.g., RSA key generation).
