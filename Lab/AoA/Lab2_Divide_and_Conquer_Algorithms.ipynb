## Related Theory (General)

**Divide and Conquer** is an algorithmic paradigm that solves a problem by breaking it into smaller sub-problems of the same type, solving each sub-problem recursively, and then combining their solutions to solve the original problem. It generally consists of three steps:

1. **Divide** — Break the problem into smaller sub-problems.
2. **Conquer** — Solve the sub-problems recursively. If the sub-problem is small enough, solve it directly (base case).
3. **Combine** — Merge/combine the solutions of the sub-problems to get the solution of the original problem.

The general recurrence relation for divide and conquer algorithms that split a problem of size $n$ into $a$ sub-problems each of size $n/b$, with $f(n)$ work done to divide/combine, is:

$$T(n) = a\,T(n/b) + f(n)$$

which can be solved using the **Master Theorem**.

## Q1. Program to Implement Binary Search

### Related Theory
**Binary Search** is a classic Divide and Conquer searching algorithm that works on a **sorted array**. It repeatedly divides the search interval in half:

- Compare the target value to the middle element of the array.
- If they are equal, the search is done.
- If the target is smaller, repeat the search on the **left half**.
- If the target is larger, repeat the search on the **right half**.
- Repeat until the element is found or the interval becomes empty.

Because the search space is halved at every step, Binary Search is much faster than a linear scan on large sorted datasets.

### Related Diagram (Flow)
```
        Start
          |
   Read sorted array, target
          |
   low = 0, high = n-1
          |
   +--------------------------------+
   | while low <= high:             | <---+
   |   mid = (low+high)//2          |     |
   |   if arr[mid] == target: found |     |
   |   elif arr[mid] < target:      |     |
   |        low = mid+1             |     |
   |   else: high = mid-1           | ----+
   +--------------------------------+
          |
   Output index (or Not Found)
          |
         End
```


````python
def binary_search(arr, target):
    """Return the index of target in sorted arr using iterative Binary Search (Divide & Conquer), or -1 if not found."""
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1


def binary_search_recursive(arr, target, low, high):
    """Recursive Divide & Conquer version of Binary Search."""
    if low > high:
        return -1
    mid = (low + high) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, high)
    else:
        return binary_search_recursive(arr, target, low, mid - 1)


# Driver / Test
if __name__ == "__main__":
    sorted_arr = [2, 5, 8, 12, 16, 23, 38, 45, 56, 72, 91]

    for target in [23, 91, 2, 100]:
        idx = binary_search(sorted_arr, target)
        if idx != -1:
            print(f"Iterative: {target} found at index {idx}")
        else:
            print(f"Iterative: {target} not found in array")

    print()
    for target in [23, 91, 2, 100]:
        idx = binary_search_recursive(sorted_arr, target, 0, len(sorted_arr) - 1)
        if idx != -1:
            print(f"Recursive: {target} found at index {idx}")
        else:
            print(f"Recursive: {target} not found in array")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(\log n)$ — the search interval is halved at every step, giving a recurrence $T(n) = T(n/2) + O(1)$, which solves to $O(\log n)$.
- **Space Complexity:** $O(1)$ for the iterative version; $O(\log n)$ for the recursive version due to the recursion call stack.
- **Remark:** Binary Search requires the input array to be **sorted**. It is far more efficient than Sequential Search ($O(n)$) for large datasets, but Sequential Search is preferred for unsorted or small data due to its simplicity and lack of a sorting precondition.

## Q2. Program to Find Min and Max of an Array using an Iterative Process

### Related Theory
The straightforward iterative approach scans the array **once**, maintaining running variables for the minimum and maximum values seen so far. At each element, both the current minimum and current maximum are compared and updated if necessary. This requires **2 comparisons per element**, giving roughly $2n$ comparisons in total.

### Related Diagram (Flow)
```
        Start
          |
   Read array of size n
          |
   min = max = arr[0]
          |
   +----------------------------------+
   | for i in range(1, n):            | <---+
   |   if arr[i] < min: min = arr[i]  |     |
   |   if arr[i] > max: max = arr[i]  | ----+
   +----------------------------------+
          |
   Output min, max
          |
         End
```


````python
def min_max_iterative(arr):
    """Return (min, max) of arr using a simple single-pass iterative scan."""
    if not arr:
        raise ValueError("Array is empty")

    minimum = maximum = arr[0]
    comparisons = 0
    for i in range(1, len(arr)):
        comparisons += 1
        if arr[i] < minimum:
            minimum = arr[i]
        comparisons += 1
        if arr[i] > maximum:
            maximum = arr[i]
    return minimum, maximum, comparisons


# Driver / Test
if __name__ == "__main__":
    test_arrays = [
        [12, 3, 45, 7, 89, 1, 34, 22],
        [5],
        [-10, -3, -45, -1, -89],
        [4, 4, 4, 4],
    ]
    for arr in test_arrays:
        mn, mx, cmp_count = min_max_iterative(arr)
        print(f"Array: {arr}\n  -> Min = {mn}, Max = {mx}, Comparisons = {cmp_count}\n")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(n)$ — a single pass through the array, with 2 comparisons per element, giving about $2(n-1)$ comparisons.
- **Space Complexity:** $O(1)$ — only two extra variables (`min`, `max`) are used.
- **Remark:** This is the simplest way to find the min and max, and is already linear (optimal order), but performs slightly more comparisons than the Divide and Conquer approach in Q3.

## Q3. Program to Find Min and Max of an Array using a Divide and Conquer Algorithm

### Related Theory
The **Divide and Conquer** approach splits the array into two halves, recursively finds the (min, max) pair of each half, and then **combines** the two pairs with just 2 comparisons:

$$\text{min} = \min(\text{min}_{left}, \text{min}_{right}), \qquad \text{max} = \max(\text{max}_{left}, \text{max}_{right})$$

Base cases:
- If there is **1 element**, it is both the min and the max (0 comparisons).
- If there are **2 elements**, one comparison determines which is min and which is max.

This reduces the total number of comparisons to about $\lceil 3n/2 \rceil - 2$, which is fewer than the $2(n-1)$ comparisons used by the naive iterative scan.

### Related Diagram (Flow)
```
        Start
          |
   MinMax(arr, low, high)
          |
   if (high - low == 0): return arr[low], arr[low]
   if (high - low == 1): return min/max of the pair (1 comparison)
          |
   mid = (low + high) // 2
   (min1, max1) = MinMax(arr, low, mid)      <- recurse left half
   (min2, max2) = MinMax(arr, mid+1, high)   <- recurse right half
          |
   min = min(min1, min2)   (Combine)
   max = max(max1, max2)   (Combine)
          |
   return min, max
          |
         End
```


````python
def min_max_divide_conquer(arr, low, high):
    """Return (min, max) of arr[low..high] using Divide & Conquer recursion."""
    # Base case: only one element
    if low == high:
        return arr[low], arr[low]

    # Base case: only two elements -> 1 comparison
    if high - low == 1:
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        else:
            return arr[high], arr[low]

    # Divide
    mid = (low + high) // 2
    min1, max1 = min_max_divide_conquer(arr, low, mid)
    min2, max2 = min_max_divide_conquer(arr, mid + 1, high)

    # Combine
    overall_min = min(min1, min2)
    overall_max = max(max1, max2)
    return overall_min, overall_max


# Driver / Test
if __name__ == "__main__":
    test_arrays = [
        [12, 3, 45, 7, 89, 1, 34, 22],
        [5],
        [-10, -3, -45, -1, -89],
        [4, 4, 4, 4],
    ]
    for arr in test_arrays:
        mn, mx = min_max_divide_conquer(arr, 0, len(arr) - 1)
        print(f"Array: {arr}\n  -> Min = {mn}, Max = {mx}\n")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(n)$, with recurrence $T(n) = 2T(n/2) + O(1)$, solved by the Master Theorem to give $\Theta(n)$. It performs about $\lceil 3n/2 \rceil - 2$ comparisons, which is fewer than the $2(n-1)$ comparisons of the plain iterative method.
- **Space Complexity:** $O(\log n)$ due to the recursion call stack (versus $O(1)$ for the iterative version).
- **Remark:** Although both approaches are asymptotically $O(n)$, the Divide and Conquer version does fewer comparisons at the cost of extra recursive call overhead and stack space — a classic **time-vs-space/overhead trade-off**.

## Q4. Program to Implement Merge Sort

### Related Theory
**Merge Sort** is a classic Divide and Conquer sorting algorithm:

1. **Divide** the array into two halves.
2. **Conquer** by recursively sorting each half.
3. **Combine** the two sorted halves using a **merge** step that repeatedly picks the smaller of the two front elements.

Merge Sort is a **stable** sort and guarantees $O(n \log n)$ performance in the best, average, and worst cases, unlike Quick Sort which can degrade to $O(n^2)$.

### Related Diagram (Flow)
<center><img src="https://www.programiz.com/sites/tutorial2program/files/merge-sort-example_0.png" width="400" alt="Description"></center>



````python
def merge_sort(arr):
    """Return a new sorted list using Merge Sort (Divide & Conquer)."""
    if len(arr) <= 1:
        return arr[:]

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])     # Divide + Conquer (left)
    right = merge_sort(arr[mid:])    # Divide + Conquer (right)
    return _merge(left, right)       # Combine


def _merge(left, right):
    """Merge two sorted lists into a single sorted list."""
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged


# Driver / Test
if __name__ == "__main__":
    test_arrays = [
        [38, 27, 43, 3, 9, 82, 10],
        [5, 2, 9, 1, 5, 6],
        [],
        [7],
        [-3, 0, -7, 4, 4, -1],
    ]
    for arr in test_arrays:
        sorted_arr = merge_sort(arr)
        print(f"Original: {arr}\nSorted  : {sorted_arr}\n")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(n \log n)$ in the **best, average, and worst** cases. The recurrence is $T(n) = 2T(n/2) + O(n)$ (the $O(n)$ term comes from the merge step), which by the Master Theorem gives $\Theta(n \log n)$.
- **Space Complexity:** $O(n)$ — auxiliary arrays are needed to hold the merged results (plus $O(\log n)$ recursion stack).
- **Remark:** Merge Sort is **stable** and has predictable performance regardless of input order, making it ideal for linked lists and external sorting (e.g., sorting data too large to fit in memory), though it uses more memory than in-place sorts like Quick Sort or Heap Sort.

## Q5. Program to Implement Heap Sort

### Related Theory
**Heap Sort** uses a **binary max-heap** data structure to sort an array in place. It works in two phases:

1. **Build a max-heap** from the input array — for every parent node, its value is ≥ the values of its children. This is done by calling `heapify` on every non-leaf node from the last one up to the root.
2. **Repeatedly extract the maximum** (the root of the heap) by swapping it with the last element of the heap, shrinking the heap size by one, and calling `heapify` again on the new root to restore the heap property. Repeating this places elements in sorted order from the end of the array backward.

While Heap Sort is not a "textbook" Divide and Conquer algorithm like Merge/Quick Sort, the `heapify` operation recursively "conquers" a subtree by comparing a node with its children (dividing the problem into the left and right subtree) and sifting the larger value down — so it shares the same recursive divide-style structure.

### Related Diagram (Flow)
```
        Start
          |
   Build max-heap from arr (heapify all internal nodes, bottom-up)
          |
   +---------------------------------------------+
   | for end in range(n-1, 0, -1):                | <---+
   |   swap arr[0] (max) with arr[end]            |     |
   |   heapify(arr, size=end, root=0)             | ----+
   +---------------------------------------------+
          |
   Output sorted array
          |
         End

   heapify(arr, n, i):
       largest = i, left = 2i+1, right = 2i+2
       if arr[left]  > arr[largest]: largest = left
       if arr[right] > arr[largest]: largest = right
       if largest != i:
           swap arr[i], arr[largest]
           heapify(arr, n, largest)   <- recurse into affected subtree
```


````python
def heapify(arr, n, i):
    """Sift the element at index i down to maintain the max-heap property within arr[0..n-1]."""
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)   # Recurse into the affected subtree


def heap_sort(arr):
    """Sort arr in place using Heap Sort. Returns the sorted array."""
    arr = arr[:]
    n = len(arr)

    # Build a max-heap (start from the last non-leaf node)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Extract elements one by one from the heap
    for end in range(n - 1, 0, -1):
        arr[0], arr[end] = arr[end], arr[0]   # move current max to the end
        heapify(arr, end, 0)                  # restore heap property on the shrunk heap

    return arr


# Driver / Test
if __name__ == "__main__":
    test_arrays = [
        [12, 11, 13, 5, 6, 7],
        [5, 2, 9, 1, 5, 6],
        [],
        [7],
        [-3, 0, -7, 4, 4, -1],
    ]
    for arr in test_arrays:
        sorted_arr = heap_sort(arr)
        print(f"Original: {arr}\nSorted  : {sorted_arr}\n")
````

### Analysis of the Algorithm
- **Time Complexity:** $O(n \log n)$ in the **best, average, and worst** cases — building the heap takes $O(n)$, and each of the $n$ extraction steps takes $O(\log n)$ for `heapify`.
- **Space Complexity:** $O(1)$ auxiliary — Heap Sort sorts **in place** (plus $O(\log n)$ recursion stack for `heapify`, or $O(1)$ if implemented iteratively).
- **Remark:** Heap Sort guarantees $O(n \log n)$ worst-case performance like Merge Sort but without needing extra array space, unlike Merge Sort. However, it is **not stable** and tends to have worse cache performance (more scattered memory access) than Quick Sort in practice.

## Q6. Program to Implement Quick Sort

### Related Theory
**Quick Sort** is a classic Divide and Conquer sorting algorithm:

1. **Divide** — Choose a **pivot** element and **partition** the array so that elements smaller than the pivot go to its left and elements larger go to its right.
2. **Conquer** — Recursively apply Quick Sort to the sub-array left of the pivot and the sub-array right of the pivot.
3. **Combine** — No explicit combine step is needed; once the sub-arrays are sorted in place, the whole array is sorted.

The efficiency of Quick Sort heavily depends on the **pivot choice**: a good pivot (close to the median) splits the array evenly, while a poor pivot (e.g., always picking the last element on an already-sorted array) leads to highly unbalanced partitions.

### Related Diagram (Flow)
```
        Start
          |
   QuickSort(arr, low, high)
          |
   if low < high:
       pivot_index = Partition(arr, low, high)   <- Divide
       QuickSort(arr, low, pivot_index - 1)      <- Conquer (left)
       QuickSort(arr, pivot_index + 1, high)     <- Conquer (right)
          |
         End

   Partition(arr, low, high):
       pivot = arr[high]
       i = low - 1
       for j in range(low, high):
           if arr[j] <= pivot:
               i += 1
               swap arr[i], arr[j]
       swap arr[i+1], arr[high]
       return i + 1
```


````python
def partition(arr, low, high):
    """Lomuto partition scheme: places pivot (last element) at its correct sorted position."""
    pivot = arr[high]
    i = low - 1  # index of the last element <= pivot

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


def quick_sort(arr, low=0, high=None):
    """Sort arr in place using Quick Sort (Divide & Conquer). Returns the sorted array."""
    if high is None:
        arr = arr[:]
        high = len(arr) - 1

    if low < high:
        pivot_index = partition(arr, low, high)      # Divide
        quick_sort(arr, low, pivot_index - 1)         # Conquer (left)
        quick_sort(arr, pivot_index + 1, high)         # Conquer (right)

    return arr


# Driver / Test
if __name__ == "__main__":
    test_arrays = [
        [10, 7, 8, 9, 1, 5],
        [5, 2, 9, 1, 5, 6],
        [],
        [7],
        [-3, 0, -7, 4, 4, -1],
    ]
    for arr in test_arrays:
        sorted_arr = quick_sort(arr)
        print(f"Original: {arr}\nSorted  : {sorted_arr}\n")
````

### Analysis of the Algorithm
- **Time Complexity:**
  - **Best/Average Case:** $O(n \log n)$ — recurrence $T(n) = 2T(n/2) + O(n)$ when the pivot splits the array roughly in half.
  - **Worst Case:** $O(n^2)$ — occurs when the pivot is always the smallest/largest element (e.g., an already-sorted array with last-element pivoting), giving recurrence $T(n) = T(n-1) + O(n)$.
- **Space Complexity:** $O(\log n)$ average (recursion stack) for in-place partitioning; $O(n)$ worst case for the recursion stack in the unbalanced case.
- **Remark:** Quick Sort is usually the **fastest sorting algorithm in practice** due to good cache locality and small constant factors, despite its $O(n^2)$ worst case. Techniques like **randomized pivot selection** or **median-of-three** pivoting are commonly used to make the worst case extremely unlikely.

## Discussion and Conclusion

In this lab, we implemented and analyzed six algorithms centered around the **Divide and Conquer** paradigm:

- **Binary Search** demonstrated the power of Divide and Conquer for searching: by discarding half of the remaining search space at every step, it achieves $O(\log n)$ performance on sorted data, far outperforming a linear scan.

It also highlighted important trade-offs among the sorting algorithms studied: Merge Sort trades space for guaranteed performance and stability, Heap Sort trades a slightly worse practical constant factor for guaranteed $O(n \log n)$ time with $O(1)$ space, and Quick Sort trades worst-case guarantees for excellent average-case practical speed.
