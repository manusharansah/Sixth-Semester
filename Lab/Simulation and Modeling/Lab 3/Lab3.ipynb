## Theory

**Monte Carlo simulation** is a computational technique that relies on repeated random sampling to
obtain numerical results for problems that might be deterministic in principle but are difficult
to solve analytically. The core idea is:

1. Define a domain of possible inputs.
2. Generate random samples from a probability distribution over that domain.
3. Perform a deterministic computation on each sample.
4. Aggregate the results.

By the **Law of Large Numbers**, as the number of random samples $n \to \infty$, the average of
the sampled results converges to the expected (true) value.

### Task 1: Estimating $\pi$
A circle of radius $r$ inscribed inside a square of side $2r$ has area $\pi r^2$, while the square
has area $(2r)^2 = 4r^2$. The ratio of areas is:

$$ \frac{\text{Area of circle}}{\text{Area of square}} = \frac{\pi r^2}{4 r^2} = \frac{\pi}{4} $$

If we drop $N$ random points uniformly inside the square and count how many, $N_{in}$, fall inside
the circle, then:

$$ \frac{N_{in}}{N} \approx \frac{\pi}{4} \quad \Longrightarrow \quad \pi \approx 4 \times \frac{N_{in}}{N} $$

### Task 2: Estimating an Average
For a set of $n$ independent random numbers $x_1, x_2, \dots, x_n$ drawn uniformly from
$[a, b]$, the sample mean $\bar{x} = \frac{1}{n}\sum x_i$ converges (by the Law of Large Numbers)
to the true expected value $\mathbb{E}[X] = \frac{a+b}{2}$ as $n$ increases.

## Task 1: Monte Carlo Estimation of $\pi$

We simulate "raindrops" falling uniformly at random inside a $1 \times 1$ square (centered at the
origin, so $x, y \in [-0.5, 0.5]$), which exactly circumscribes a circle of radius $0.5$. Each drop
is classified as landing inside or outside the circle based on whether $x^2 + y^2 \le 0.5^2$.


```
import numpy as np
import random
import math
import matplotlib.pyplot as plt

random.seed(42)  # for reproducibility; remove/change the seed to get a different random run

def monte_carlo_pi(number_of_drops):
    """Simulate dropping points into a unit square with an inscribed circle of radius 0.5.
    Returns (count_inside, total_drops, x_in, y_in, x_out, y_out, pi_history)."""
    x_in, y_in = [], []
    x_out, y_out = [], []
    count_in = 0
    pi_history = []

    for i in range(number_of_drops):
        x = random.uniform(-0.5, 0.5)
        y = random.uniform(-0.5, 0.5)

        if x**2 + y**2 <= 0.5**2:
            count_in += 1
            x_in.append(x)
            y_in.append(y)
        else:
            x_out.append(x)
            y_out.append(y)

        pi_history.append(4 * count_in / (i + 1))

    return count_in, number_of_drops, x_in, y_in, x_out, y_out, pi_history
```

```
number_of_drops = int(input("Enter the number of drops: "))

count_in, total, x_in, y_in, x_out, y_out, pi_history = monte_carlo_pi(number_of_drops)
result = [count_in, total]

print("-----------------------")
print(f"{number_of_drops} drops")
print(f"π estimated as: {4 * result[0] / result[1]:.6f}")
print(f"Actual π: {math.pi:.6f}")
print("-----------------------")
```

```
plt.figure(figsize=(6, 6))
plt.scatter(x_in, y_in, color='blue', s=10, label='Drops in circle')
plt.scatter(x_out, y_out, color='black', s=10, label='Drops out of circle')
plt.legend()
plt.title(f"{number_of_drops} drops: {count_in} landed in circle, estimating π as {4*count_in/number_of_drops:.4f}.")
plt.gca().set_aspect('equal')
plt.show()
```

```
plt.figure(figsize=(8, 5))
plt.plot(range(1, number_of_drops + 1), pi_history, color='tab:blue')
plt.axhline(y=math.pi, color='black', linestyle='--', label='Actual π')
plt.xlabel('Number of rain drops')
plt.ylabel('π')
plt.title('π estimate against number of rain drops')
plt.legend()
plt.show()
```

## Task 2: Monte Carlo Estimation of an Average

The Monte Carlo simulation is used to estimate the average value of randomly generated integers
between 1 and 9 (`np.random.randint(1, 10)` generates integers in $[1, 9]$, so the expected/true
average is $5.0$). For each epoch $i$, we generate $i+1$ random numbers and compute their running
average, and plot how this average converges to the true expected value as the sample size grows.


```
import numpy as np
import matplotlib.pyplot as plt

EPOCH = 10000
AVG = []

for i in range(EPOCH):
    add = []
    for j in range(i + 1):
        add.append(np.random.randint(1, 10))
    avg = np.mean(add)
    AVG.append(avg)

plt.figure(figsize=(9, 6))
plt.plot(AVG, color='y')
plt.title('Average between 1-10 using MonteCarlo [ACE080BCT037]')
plt.xlabel('Number of epoch(s)')
plt.ylabel('Numbers (1-10)')
plt.axhline(y=5.0, color='r', alpha=0.4, linestyle='--')
plt.show()
```
